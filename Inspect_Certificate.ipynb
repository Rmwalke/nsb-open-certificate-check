# Inspect a finite NS-B certificate
Run the two code cells in order. The script recomputes exact-rational invariant-set and third-iterate-entry inequalities using stored enclosure inputs. A pass is conditional on those inputs; it is not a verification of the entire mathematical proof.

Paper author: Rushikesh Murlidhar Walke. DOI: https://doi.org/10.5281/zenodo.23429477

This notebook uses only Python standard-library modules. The local calculation was tested; a public Binder launch has not been tested.

```python
from pathlib import Path
import subprocess, sys, json
root = Path.cwd()
run = subprocess.run([sys.executable, str(root/"audit/check_certificate.py"), "--code-dir", str(root/"Code"), "--output-dir", str(root/"audit/run_results")], capture_output=True, text=True)
print(run.stdout)
if run.returncode:
    print(run.stderr)
    raise RuntimeError("The limited certificate check did not pass; inspect the diagnostics.")
```

```python
result = json.loads((root/"audit/run_results/W15_data_only_audit.json").read_text())
print("Scope:", result["scope"])
print("Block | invariant condition | third-iterate entry | entry ratio (display only)")
for row in result["blocks"]:
    print(row["block"], row["invariance_pass"], row["fresh_entry_pass"], row["fresh_entry_ratio_display"])
print("\nLimits:")
for note in result["assumptions_and_limits"]:
    print("-", note)
```
